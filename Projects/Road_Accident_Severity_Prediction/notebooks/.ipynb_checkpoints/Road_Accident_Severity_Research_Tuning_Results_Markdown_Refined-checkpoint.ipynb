# Predicting Road Accident Severity Using Machine Learning

## Project Overview

This project examines road traffic accident records from Addis Ababa (2017–2020) to explore patterns associated with Slight, Serious, and Fatal Injury outcomes. Logistic Regression, Decision Tree, Random Forest, and XGBoost are compared under balanced and unbalanced training conditions. The analysis is retrospective: it classifies severity in recorded incidents and does not establish causes or provide a validated real-time risk prediction system.

## Business Problem

Road accidents create serious public safety challenges, but identifying the characteristics associated with more severe outcomes is difficult. A model that mainly recognizes common Slight Injury cases could appear accurate while missing Serious or Fatal Injury cases. This study examines whether historical accident characteristics can help distinguish severity levels and what limitations must be considered before such results inform road-safety decisions.

## Research Questions

1. Which recorded accident characteristics are most influential in the selected classification model?
2. What relationship exists between weather conditions and recorded accident severity?
3. How are time-related and road characteristics associated with accident severity?
4. How well can machine-learning models classify Slight, Serious, and Fatal Injury cases, particularly the minority classes?
5. Which of the four algorithms and training configurations performs best under the chosen model-selection criterion?

## Dataset Description

The Road Traffic Severity Classification dataset contains **12,316 recorded accidents** from Addis Ababa, Ethiopia, covering **2017–2020**. The target variable, `Accident_severity`, has three categories: Slight Injury, Serious Injury, and Fatal Injury. The dataset contains driver, vehicle, road, time, weather, and collision information. Because the severity distribution is highly imbalanced, model assessment includes Macro F1, balanced accuracy, and class-level recall rather than accuracy alone.

**Dataset source:** [Road Traffic Severity Classification (Kaggle)](https://www.kaggle.com/datasets/avikumart/road-traffic-severity-classification).

## 6. Creating Time-Based Features

The recorded `Time` field is converted into a datetime representation and used to derive the accident hour. A second feature groups hours into Morning (05:00–11:59), Afternoon (12:00–16:59), Evening (17:00–20:59), and Night (21:00–04:59). Missing or unparseable times are labeled `Unknown` for the grouped feature. These derived variables support the research question about temporal associations; they do not establish that the time of day causes accident severity.

```python
# Convert accident time and create more useful time-based features.
df["Time"] = pd.to_datetime(
    df["Time"],
    format="%H:%M:%S",
    errors="coerce"
)

df["Hour"] = df["Time"].dt.hour


def classify_time(hour):

    if pd.isna(hour):
        return "Unknown"

    elif 5 <= hour < 12:
        return "Morning"

    elif 12 <= hour < 17:
        return "Afternoon"

    elif 17 <= hour < 21:
        return "Evening"

    else:
        return "Night"


df["Time_of_day"] = df["Hour"].apply(classify_time)
```

## 14. Measuring Statistical Associations

Chi-square tests assess statistical associations between categorical accident characteristics and severity, while Cramér’s V measures association strength. Statistical significance is considered separately from practical importance; neither measure establishes causation.

```python
# Chi-square identifies statistical association.
# Cramer's V measures the strength of that association.

def cramers_v(x, y):

    table = pd.crosstab(x, y)

    chi2, p, dof, expected = chi2_contingency(table)

    n = table.values.sum()

    phi2 = chi2 / n

    r, k = table.shape

    phi2corr = max(
        0,
        phi2 - ((k - 1) * (r - 1)) / (n - 1)
    )

    rcorr = r - (
        ((r - 1) ** 2) / (n - 1)
    )

    kcorr = k - (
        ((k - 1) ** 2) / (n - 1)
    )

    denominator = min(
        kcorr - 1,
        rcorr - 1
    )

    if denominator <= 0:
        return 0

    return np.sqrt(
        phi2corr / denominator
    )


association_features = [
    "Weather_conditions",
    "Light_conditions",
    "Road_surface_conditions",
    "Time_of_day",
    "Day_of_week",
    "Road_allignment",
    "Types_of_Junction",
    "Road_surface_type",
    "Type_of_collision"
]


association_results = []


for feature in association_features:

    temp = df[
        [
            feature,
            "Accident_severity"
        ]
    ].dropna()

    contingency_table = pd.crosstab(
        temp[feature],
        temp["Accident_severity"]
    )

    chi2, p_value, dof, expected = (
        chi2_contingency(
            contingency_table
        )
    )

    cv = cramers_v(
        temp[feature],
        temp["Accident_severity"]
    )

    association_results.append({
        "Feature": feature,
        "Chi-Square": chi2,
        "P-Value": p_value,
        "Cramers V": cv
    })


association_df = (
    pd.DataFrame(
        association_results
    )
    .sort_values(
        by="Cramers V",
        ascending=False
    )
)

display(
    association_df.round(4)
)
```

### Statistical Association Results

**Time of Day** has the highest Cramér's V in this table (**0.0441**), followed by **Types of Junction** (**0.0419**). Even these leading values indicate very weak individual associations with accident severity. Several p-values are statistically significant, but significance does not mean the relationship is strong or causal. These tests examine each variable separately and do not account for interactions with other accident characteristics.

## 15. Preparing Data for Modeling

Preparing the predictors includes checking for variables that could reveal the outcome. `Number_of_casualties` is excluded because it describes accident consequences, while `Cause_of_accident` may only be determined after investigation. The resulting feature set is used consistently across model comparisons.

```python
df_model = df.copy()

df_model = df_model.dropna(
    subset=["Accident_severity"]
)

# Remove the target variable and variables that may introduce
# target leakage or represent post-accident casualty information.
# The original Time variable is removed because Hour and
# Time_of_day were already created during feature engineering.

columns_to_drop = [
    "Accident_severity",
    "Casualty_severity",
    "Casualty_class",
    "Sex_of_casualty",
    "Age_band_of_casualty",
    "Work_of_casuality",
    "Fitness_of_casuality",
    "Time",
    "Number_of_casualties",
    "Cause_of_accident"
]

X = df_model.drop(
    columns=columns_to_drop,
    errors="ignore"
)

# Standardize inconsistent case and surrounding spaces in the severity labels.
# Example: "Fatal injury " becomes "Fatal Injury".
y = df_model["Accident_severity"].astype(str).str.strip().str.title()
expected_classes = {"Slight Injury", "Serious Injury", "Fatal Injury"}
observed_classes = set(y.unique())
if observed_classes != expected_classes:
    raise ValueError(
        f"Unexpected accident severity labels: {sorted(observed_classes)}. "
        f"Expected: {sorted(expected_classes)}"
    )


# Encode the three accident-severity classes as numeric labels
# so they can be used by the classification models.
label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)


print("Target Classes:")

for number, label in enumerate(
    label_encoder.classes_
):
    print(
        number,
        "=",
        label
    )

print("\nNumber of Predictor Variables:", X.shape[1])
```

### Predictor Availability and Leakage Check

The predictor availability review distinguishes usable predictors from variables excluded because of possible information leakage. A retrospective classification dataset can include information that would not be known before an investigation.

```python
print("Number_of_casualties included?", "Number_of_casualties" in X.columns)
print("Cause_of_accident included?", "Cause_of_accident" in X.columns)
print("Remaining predictor columns:", X.columns.tolist())
assert "Number_of_casualties" not in X.columns
assert "Cause_of_accident" not in X.columns
```

### Leakage Check

`Number_of_casualties` was excluded because it describes the consequences of an accident and could reveal information closely related to the target. `Cause_of_accident` was also excluded as a conservative decision because the official cause may be assigned only after investigation. The predictor check above verifies whether these fields remain in `X`. Other fields, including collision type, should still be reviewed against the intended prediction time. Avoiding leakage is important because a model can score well during testing if it has access to information that would not exist when used in practice.

## 16. Creating Training and Test Sets

A stratified 80/20 split preserves the relative proportions of Slight, Serious, and Fatal Injury cases across the training and held-out test partitions. Training data are used for fitting preprocessing, cross-validation, and tuning. The test partition is used to evaluate model performance and is not used to choose hyperparameters.

```python
# Use stratification to preserve the accident-severity
# distribution in both the training and test datasets.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y_encoded
)

print("Training Records:", X_train.shape[0])
print("Testing Records:", X_test.shape[0])
```

### Training and Test Split — Interpretation

The recorded split produced **9,852 training records** and **2,464 test records**. Stratification is particularly important here because Fatal Injury cases represent a small minority of observations. The printed record counts above should be checked if the dataset or split settings change.

## 17. Building the Preprocessing Pipeline

Numerical predictors receive median imputation followed by standardization. Categorical predictors receive most-frequent-value imputation and one-hot encoding; infrequent categories are grouped using `min_frequency=10`, and previously unseen categories are handled by the encoder. The `ColumnTransformer` applies these steps consistently to each model. Because preprocessing is fitted inside each training fold, validation and test records do not determine the imputation or encoding parameters. This cell defines the pipeline; it does not produce performance metrics.

```python
numeric_features = (
    X_train
    .select_dtypes(include=np.number)
    .columns
    .tolist()
)

categorical_features = (
    X_train
    .select_dtypes(exclude=np.number)
    .columns
    .tolist()
)


numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)


categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="infrequent_if_exist",
                min_frequency=10
            )
        )
    ]
)


preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_transformer,
            numeric_features
        ),
        (
            "categorical",
            categorical_transformer,
            categorical_features
        )
    ]
)
```

## 18. Defining Four Algorithms Under Two Training Conditions

Four algorithms represent different classification approaches: Logistic Regression (linear baseline), Decision Tree (single nonlinear tree), Random Forest (bagged ensemble), and XGBoost (gradient-boosted trees). Each is evaluated in **unbalanced** and **balanced** conditions, yielding eight configurations. Logistic Regression, Decision Tree, and Random Forest use `class_weight="balanced"` in their balanced versions. Balanced XGBoost instead receives training-derived sample weights through `compute_sample_weight`. The configurations share the same data split and preprocessing approach, allowing comparison of class-level recall, Macro F1, balanced accuracy, and overall accuracy. This section defines the experiments; their observed performance is interpreted in the later results sections.

```python
# Eight comparable configurations: four algorithms x two weighting conditions.
models = {
    "Logistic Regression | Unbalanced": LogisticRegression(max_iter=2000, class_weight=None, random_state=RANDOM_STATE),
    "Logistic Regression | Balanced": LogisticRegression(max_iter=2000, class_weight="balanced", random_state=RANDOM_STATE),
    "Decision Tree | Unbalanced": DecisionTreeClassifier(max_depth=12, min_samples_leaf=5, class_weight=None, random_state=RANDOM_STATE),
    "Decision Tree | Balanced": DecisionTreeClassifier(max_depth=12, min_samples_leaf=5, class_weight="balanced", random_state=RANDOM_STATE),
    "Random Forest | Unbalanced": RandomForestClassifier(n_estimators=300, min_samples_leaf=2, class_weight=None, random_state=RANDOM_STATE, n_jobs=-1),
    "Random Forest | Balanced": RandomForestClassifier(n_estimators=300, min_samples_leaf=2, class_weight="balanced", random_state=RANDOM_STATE, n_jobs=-1),
    "XGBoost | Unbalanced": XGBClassifier(n_estimators=300, max_depth=6, learning_rate=0.05, subsample=0.8, colsample_bytree=0.8, objective="multi:softprob", eval_metric="mlogloss", random_state=RANDOM_STATE, n_jobs=-1),
    "XGBoost | Balanced": XGBClassifier(n_estimators=300, max_depth=6, learning_rate=0.05, subsample=0.8, colsample_bytree=0.8, objective="multi:softprob", eval_metric="mlogloss", random_state=RANDOM_STATE, n_jobs=-1),
}

def fit_comparison_pipeline(pipeline, X_fit, y_fit, name):
    if name == "XGBoost | Balanced":
        weights = compute_sample_weight(class_weight="balanced", y=y_fit)
        pipeline.fit(X_fit, y_fit, model__sample_weight=weights)
    else:
        pipeline.fit(X_fit, y_fit)
    return pipeline
```

## 19. Training and Evaluating Models

Eight model configurations are trained and evaluated using the same held-out test set. In addition to accuracy, Macro F1, balanced accuracy, and recall for Serious and Fatal Injury are reported because minority-class performance matters for this dataset.

The following code fits each configuration and prints a classification report for each severity class. The interpretation immediately after the code summarizes the **saved numerical results**, rather than simply describing the procedure.

```python
# Store model results, fitted pipelines, and predictions.
results = []
fitted_models = {}
predictions = {}


# Train and evaluate each model using the same preprocessing
# pipeline and train/test data for a consistent comparison.
for model_name, model in models.items():

    print("\n" + "=" * 60)
    print("Training:", model_name)
    print("=" * 60)

    # Keep preprocessing and modeling in one pipeline.
    pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", model)
        ]
    )

    # Train model
    fit_comparison_pipeline(pipeline, X_train, y_train, model_name)

    # Make predictions
    y_pred = pipeline.predict(
        X_test
    )

    # Store fitted model
    fitted_models[model_name] = pipeline

    # Store predictions for later evaluation
    predictions[model_name] = y_pred

    # Macro metrics give equal importance to each severity class.
    # This is useful because the target classes are highly imbalanced.
    results.append({

        "Model":
            model_name,

        "Accuracy":
            accuracy_score(
                y_test,
                y_pred
            ),

        "Balanced Accuracy":
            balanced_accuracy_score(
                y_test,
                y_pred
            ),

        "Macro Precision":
            precision_score(
                y_test,
                y_pred,
                average="macro",
                zero_division=0
            ),

        "Macro Recall":
            recall_score(
                y_test,
                y_pred,
                average="macro",
                zero_division=0
            ),

        "Macro F1":
            f1_score(
                y_test,
                y_pred,
                average="macro",
                zero_division=0
            ),

        "Weighted F1":
            f1_score(
                y_test,
                y_pred,
                average="weighted",
                zero_division=0
            )
    })

    # Display class-level precision, recall, and F1 results.
    print("\nClassification Report:\n")

    print(
        classification_report(
            y_test,
            y_pred,
            target_names=label_encoder.classes_,
            zero_division=0
        )
    )


# Convert model results to a DataFrame.
results_df = pd.DataFrame(
    results
)

# Rank models by Macro F1 because it gives equal weight
# to Fatal, Serious, and Slight Injury classes.
results_df = results_df.sort_values(
    by="Macro F1",
    ascending=False
).reset_index(
    drop=True
)


print("\nModel Comparison:\n")

display(
    results_df.round(4)
)
```

### Training and Evaluation Results — Interpretation

The initial results demonstrate a substantial trade-off between overall accuracy and recognition of minority severity classes. **Unbalanced Logistic Regression** obtained **84.62% accuracy** but **0% Fatal Injury recall**, while **Balanced Logistic Regression** obtained **47.85% accuracy** and **38.71% Fatal Injury recall**. **Unbalanced Random Forest** also had **0% Fatal Injury recall** despite **84.74% accuracy**. These figures show why the classification reports must be read alongside accuracy. The complete eight-model comparison is summarized in the next section; final selection is based on **training cross-validation**, not these held-out test outcomes.

## 20. Machine Learning Model Performance

## 21. Performing Cross-Validation

Five-fold stratified cross-validation assesses model performance across multiple training/validation splits. Balanced and unbalanced XGBoost configurations are kept separate so that cross-validation reflects their respective training approaches.

```python
# Five-fold stratified validation helps determine
# whether performance is consistent across training subsets.

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=RANDOM_STATE
)


cv_results = []


for model_name, model in models.items():

    fold_scores = []


    for train_index, validation_index in cv.split(
        X_train,
        y_train
    ):

        X_cv_train = X_train.iloc[
            train_index
        ]

        X_cv_validation = X_train.iloc[
            validation_index
        ]

        y_cv_train = y_train[
            train_index
        ]

        y_cv_validation = y_train[
            validation_index
        ]


        cv_pipeline = Pipeline(
            steps=[
                (
                    "preprocessor",
                    preprocessor
                ),
                (
                    "model",
                    model
                )
            ]
        )


        # Match the baseline held-out model configuration exactly.
        fit_comparison_pipeline(cv_pipeline, X_cv_train, y_cv_train, model_name)

        cv_prediction = (
            cv_pipeline.predict(
                X_cv_validation
            )
        )


        fold_scores.append(
            f1_score(
                y_cv_validation,
                cv_prediction,
                average="macro",
                zero_division=0
            )
        )


    cv_results.append({

        "Model":
            model_name,

        "CV Macro F1":
            np.mean(
                fold_scores
            )
    })


cv_results_df = (
    pd.DataFrame(
        cv_results
    )
    .sort_values(
        by="CV Macro F1",
        ascending=False
    )
)


display(
    cv_results_df.round(4)
)
```

### Comparing Balanced and Unbalanced Models

The next table combines the initial held-out test metrics with **five-fold training cross-validation Macro F1** and class-specific recall. Balanced training here means **class weighting** (or **sample weighting for XGBoost**). The table permits within-algorithm comparisons without treating the test set as the model-selection criterion.

```python
# One comparison table covering all four algorithms and both conditions.
all_model_comparison = results_df.merge(cv_results_df, on="Model", how="left")
for name, prediction in predictions.items():
    class_recalls = recall_score(y_test, prediction, labels=np.arange(len(label_encoder.classes_)), average=None, zero_division=0)
    for label, recall_value in zip(label_encoder.classes_, class_recalls):
        all_model_comparison.loc[all_model_comparison["Model"] == name, f"{label} Recall"] = recall_value
all_model_comparison["Algorithm"] = all_model_comparison["Model"].str.split(" | ", regex=False).str[0]
all_model_comparison["Training"] = all_model_comparison["Model"].str.split(" | ", regex=False).str[-1]
all_model_comparison = all_model_comparison.sort_values(["Algorithm", "Training"])
display(all_model_comparison.round(4))
all_model_comparison.to_csv(RESULTS_DIR / "all_eight_model_comparison.csv", index=False)
```

In the baseline test results, **Balanced XGBoost** improved Macro F1 from **0.3340 to 0.3967** compared with Unbalanced XGBoost, but accuracy declined from **84.74% to 72.32%**. Its **Fatal Injury recall was 9.68%**, versus **0%** for the unbalanced version; Serious Injury recall increased from **4.58% to 36.10%**.

For **Random Forest**, balancing the untuned model produced only a small Macro F1 increase (**0.3133 to 0.3226**), and both configurations had **0% Fatal Injury recall**. For **Logistic Regression**, balancing increased Fatal Injury recall from **0% to 38.71%**, but reduced overall accuracy from **84.62% to 47.85%**, showing the trade-off between minority recall and false alarms. None of these results makes the baseline models suitable for safety-critical use.

### Cross-Validation and Test Results — Interpretation

The **training cross-validation Macro F1** was highest for **Balanced XGBoost (0.4235)** among the untuned configurations, compared with **0.3205** for Balanced Random Forest. These training-fold results are distinct from the held-out test metrics shown in the same table. With only **31 Fatal Injury cases** in the test partition, minority-class recall estimates are particularly uncertain.

## 22. Hyperparameter Tuning for All Eight Model Configurations

Tuned balanced and unbalanced versions are compared of Logistic Regression, Decision Tree, Random Forest, and XGBoost using the **same five-fold stratified cross-validation** and **Macro F1** as the selection metric. The existing untuned results remain above as a reference.

- Hyperparameter combinations are sampled with `ParameterSampler` using a fixed random seed. Each candidate is evaluated only on training folds; preprocessing was fitted inside each fold.
- Logistic Regression, Decision Tree, and Random Forest use built-in `class_weight` for the balanced variants. Balanced XGBoost uses `compute_sample_weight` **within each training fold**.
- All eight configurations use identical train/test partitions and CV folds. Their hyperparameter spaces are appropriate to each algorithm but are not identical because their parameters differ.
- The test set is **not used for hyperparameter selection**. Five-fold CV on the training portion chooses the parameters; final held-out test metrics are descriptive, not a second tuning criterion.
- This can take substantial time, especially for XGBoost and Random Forest. Reduce `N_SEARCH_ITER` temporarily for a smoke test, then restore it for final results. Parallelism is limited to avoid oversubscribing the machine.

Hyperparameters are selected using five-fold cross-validation on the training data. Because the same folds are used to compare candidates, the winning CV score may be somewhat optimistic. Held-out test performance provides a separate evaluation of the selected configuration.

```python
from sklearn.model_selection import ParameterSampler
from sklearn.base import clone
from sklearn.metrics import roc_auc_score

# Up to 10 combinations per configuration (5 for Logistic Regression), 5 folds each.
N_SEARCH_ITER = 10
TUNING_FOLDS = 5
TUNING_SEED = RANDOM_STATE

tuning_cv = StratifiedKFold(n_splits=TUNING_FOLDS, shuffle=True, random_state=RANDOM_STATE)

# Keys are model parameters (not pipeline parameters); each candidate is
# installed in the 'model' step of the existing preprocessing pipeline.
param_spaces = {
    "Logistic Regression": {
        "C": [0.01, 0.1, 1.0, 10.0, 100.0],
        "solver": ["lbfgs"],
    },
    "Decision Tree": {
        "max_depth": [4, 8, 12, 18, None],
        "min_samples_split": [2, 5, 10, 20],
        "min_samples_leaf": [1, 2, 5, 10],
    },
    "Random Forest": {
        "n_estimators": [100, 200, 300],
        "max_depth": [8, 12, 20, None],
        "min_samples_leaf": [1, 2, 5],
        "max_features": ["sqrt", 0.75],
    },
    "XGBoost": {
        "n_estimators": [150, 250, 350],
        "max_depth": [3, 5, 7],
        "learning_rate": [0.03, 0.05, 0.1],
        "subsample": [0.8, 1.0],
        "colsample_bytree": [0.8, 1.0],
    },
}

# Use the same CV folds for every configuration. CV is restricted to X_train.
fold_indices = list(tuning_cv.split(X_train, y_train))
tuning_search_rows = []
tuned_cv_rows = []
tuned_models = {}
tuned_predictions = {}
tuned_test_rows = []

for config_name, base_model in models.items():
    family = config_name.split(" | ")[0]
    candidates = list(ParameterSampler(
        param_spaces[family], n_iter=N_SEARCH_ITER, random_state=TUNING_SEED
    ))
    best_mean = -np.inf
    best_params = None
    best_fold_scores = None
    print("Tuning:", config_name, "| candidates:", len(candidates))

    for candidate_no, params in enumerate(candidates, start=1):
        fold_f1 = []
        for train_idx, val_idx in fold_indices:
            # New estimator and preprocessing pipeline per fold: no fitted-state leakage.
            estimator = clone(base_model).set_params(**params)
            if hasattr(estimator, "n_jobs"):
                estimator.set_params(n_jobs=2)
            candidate_pipeline = Pipeline([
                ("preprocessor", clone(preprocessor)),
                ("model", estimator),
            ])
            X_fold_train, X_fold_val = X_train.iloc[train_idx], X_train.iloc[val_idx]
            y_fold_train, y_fold_val = y_train[train_idx], y_train[val_idx]
            fit_comparison_pipeline(candidate_pipeline, X_fold_train, y_fold_train, config_name)
            pred = candidate_pipeline.predict(X_fold_val)
            fold_f1.append(f1_score(y_fold_val, pred, average="macro", zero_division=0))

        mean_f1 = float(np.mean(fold_f1))
        tuning_search_rows.append({
            "Model": config_name, "Candidate": candidate_no,
            "CV Macro F1": mean_f1, "CV Macro F1 SD": float(np.std(fold_f1)),
            "Parameters": str(params),
        })
        if mean_f1 > best_mean:
            best_mean = mean_f1
            best_params = params.copy()
            best_fold_scores = fold_f1.copy()

    tuned_cv_rows.append({
        "Model": config_name,
        "Tuned CV Macro F1": best_mean,
        "Tuned CV Macro F1 SD": float(np.std(best_fold_scores)),
        "Best Parameters": str(best_params),
    })

    # Fit best training-CV configuration once on the entire training partition.
    final_estimator = clone(base_model).set_params(**best_params)
    if hasattr(final_estimator, "n_jobs"):
        final_estimator.set_params(n_jobs=2)
    final_pipeline = Pipeline([
        ("preprocessor", clone(preprocessor)),
        ("model", final_estimator),
    ])
    fit_comparison_pipeline(final_pipeline, X_train, y_train, config_name)
    tuned_models[config_name] = final_pipeline
    test_pred = final_pipeline.predict(X_test)
    tuned_predictions[config_name] = test_pred
    class_recalls = recall_score(y_test, test_pred, average=None, labels=np.arange(len(label_encoder.classes_)), zero_division=0)
    row = {
        "Model": config_name,
        "Accuracy": accuracy_score(y_test, test_pred),
        "Balanced Accuracy": balanced_accuracy_score(y_test, test_pred),
        "Macro F1": f1_score(y_test, test_pred, average="macro", zero_division=0),
        "Weighted F1": f1_score(y_test, test_pred, average="weighted", zero_division=0),
    }
    for label_name, recall_value in zip(label_encoder.classes_, class_recalls):
        row[f"Recall - {label_name}"] = recall_value
    tuned_test_rows.append(row)

# Keep all candidates for transparency and reproducibility.
tuning_search_df = pd.DataFrame(tuning_search_rows)
tuned_cv_df = pd.DataFrame(tuned_cv_rows).sort_values("Tuned CV Macro F1", ascending=False)
tuned_test_df = pd.DataFrame(tuned_test_rows).sort_values("Macro F1", ascending=False)
print("\nBest training-CV parameters (ranked by CV Macro F1):")
display(tuned_cv_df.round(4))
print("\nHeld-out test results for the tuned configurations:")
display(tuned_test_df.round(4))
```

### Comparing Untuned and Tuned Models

Untuned and tuned cross-validation Macro F1 scores are compared for each configuration. Because tuning selects the best candidate, the tuned cross-validation score is used for model selection rather than treated as an unbiased estimate of future performance. The selected tuned Balanced Random Forest is then evaluated separately on the held-out test set.

```python
# Summary: compare like-for-like configurations across untuned and tuned runs.
untuned_test = results_df[["Model", "Accuracy", "Balanced Accuracy", "Macro F1"]].rename(columns={
    "Accuracy": "Untuned Test Accuracy",
    "Balanced Accuracy": "Untuned Test Balanced Accuracy",
    "Macro F1": "Untuned Test Macro F1",
})
untuned_cv = cv_results_df[["Model", "CV Macro F1"]].rename(columns={
    "CV Macro F1": "Untuned CV Macro F1",
})
tuned_test = tuned_test_df[["Model", "Accuracy", "Balanced Accuracy", "Macro F1"]].rename(columns={
    "Accuracy": "Tuned Test Accuracy",
    "Balanced Accuracy": "Tuned Test Balanced Accuracy",
    "Macro F1": "Tuned Test Macro F1",
})
comparison_tuned_df = (
    untuned_test.merge(untuned_cv, on="Model")
    .merge(tuned_cv_df[["Model", "Tuned CV Macro F1"]], on="Model")
    .merge(tuned_test, on="Model")
)
comparison_tuned_df["Change in Test Macro F1"] = (
    comparison_tuned_df["Tuned Test Macro F1"] - comparison_tuned_df["Untuned Test Macro F1"]
)
display(comparison_tuned_df.sort_values("Tuned CV Macro F1", ascending=False).round(4))

# Select using TRAINING CV ONLY, not held-out test performance.
selected_config = tuned_cv_df.iloc[0]["Model"]
print("Selected by tuned training CV Macro F1:", selected_config)
print("Held-out test metrics for selected model:")
display(tuned_test_df.loc[tuned_test_df["Model"] == selected_config].round(4))

# Save separately; do not overwrite the original model_comparison.csv.
tuning_search_df.to_csv(RESULTS_DIR / "tuning_all_candidates.csv", index=False)
tuned_cv_df.to_csv(RESULTS_DIR / "tuned_cv_results.csv", index=False)
tuned_test_df.to_csv(RESULTS_DIR / "tuned_test_results.csv", index=False)
comparison_tuned_df.to_csv(RESULTS_DIR / "untuned_vs_tuned_comparison.csv", index=False)
print("Tuning comparison CSV files saved in results/ (these are artifacts, not performance metrics).")
```

### Tuned Model Results

The tuning results report both the training cross-validation scores used to select model settings and the held-out test scores used for evaluation. The **Balanced Random Forest** achieved the highest tuned CV Macro F1 (**0.4295**) and was selected for final evaluation. Its test Macro F1 was **0.4193**, with balanced accuracy of **0.4184**. The low Fatal Injury recall remains an important limitation despite the model's higher overall accuracy.

### Hyperparameter Tuning Results — Interpretation

The tuning search uses **five-fold stratified cross-validation on the training set**. Logistic Regression has **five candidates**, because its search space contains five `C` values and one fixed solver (`lbfgs`); there are only five unique combinations. The other algorithm families have larger search spaces, so **ten randomly sampled combinations** are evaluated per balanced or unbalanced configuration. This is a computational budget, not evidence that ten combinations are optimal.

In the recorded tuning results, **Balanced Random Forest** achieved the highest **training CV Macro F1 (0.4295)**, narrowly ahead of **Balanced XGBoost (0.4236)**. Its held-out test **Macro F1 was 0.4193**, versus **0.4093** for Balanced XGBoost. Tuning raised Balanced Random Forest test Macro F1 from **0.3226 to 0.4193**, but its Fatal Injury recall remained only **9.68%**. These are observed results from the saved execution, not a guarantee of improvement on new datasets.

## 24. Ethical Evaluation: Subgroup Performance Review

A concrete performance audit examines whether the selected tuned model performs differently across **lighting, weather, and road-surface conditions**. The audit uses the same held-out test predictions as the final model evaluation. For each subgroup, it reports sample size, accuracy, Macro F1, and recall for Serious and Fatal Injury when those classes are represented.

Small subgroups and rare Fatal Injury cases can produce unstable estimates. Missing class examples are shown as unavailable rather than assigned a misleading zero. These are **operational subgroups**, not demographic fairness groups, and the results do not establish equal treatment across people or populations.

```python
# Ethical evaluation: performance consistency across accident conditions.
# This is a descriptive audit on the held-out test set, not model selection.
from sklearn.metrics import accuracy_score, f1_score, recall_score

severity_names = list(label_encoder.classes_)
# Use the labels fitted during target preparation; avoid silently assuming
# that raw dataset spellings/capitalization match the intended categories.
required_labels = {"Slight Injury", "Serious Injury", "Fatal Injury"}
if set(severity_names) != required_labels:
    raise ValueError(f"Unexpected encoded severity labels: {severity_names}")
serious_id = int(label_encoder.transform(["Serious Injury"])[0])
fatal_id = int(label_encoder.transform(["Fatal Injury"])[0])

audit_frame = X_test.copy()
audit_frame["Actual Severity"] = np.asarray(y_test)
audit_frame["Predicted Severity"] = np.asarray(best_prediction)

subgroup_fields = {
    "Lighting": "Light_conditions",
    "Weather": "Weather_conditions",
    "Road Surface": "Road_surface_conditions",
}

subgroup_rows = []
for category, field in subgroup_fields.items():
    if field not in audit_frame.columns:
        print(f"Skipping {category}: variable not present in held-out data.")
        continue
    for group_value, group in audit_frame.groupby(field, dropna=False):
        actual = group["Actual Severity"].to_numpy()
        predicted = group["Predicted Severity"].to_numpy()
        serious_count = int(np.sum(actual == serious_id))
        fatal_count = int(np.sum(actual == fatal_id))
        serious_recall = (
            float(np.mean(predicted[actual == serious_id] == serious_id))
            if serious_count > 0 else np.nan
        )
        fatal_recall = (
            float(np.mean(predicted[actual == fatal_id] == fatal_id))
            if fatal_count > 0 else np.nan
        )
        subgroup_rows.append({
            "Condition": category,
            "Group": "Missing" if pd.isna(group_value) else str(group_value),
            "Records": len(group),
            "Serious Cases": serious_count,
            "Fatal Cases": fatal_count,
            "Accuracy": accuracy_score(actual, predicted),
            "Macro F1": f1_score(actual, predicted, labels=np.arange(len(severity_names)),
                                  average="macro", zero_division=0),
            "Serious Recall": serious_recall,
            "Fatal Recall": fatal_recall,
        })

subgroup_audit_df = pd.DataFrame(subgroup_rows)
subgroup_audit_df = subgroup_audit_df.sort_values(
    ["Condition", "Records"], ascending=[True, False]
).reset_index(drop=True)

# Avoid interpreting rare groups as reliable evidence; retain them transparently.
MIN_GROUP_RECORDS = 30
MIN_CLASS_CASES = 5
subgroup_audit_df["Adequate Group Size"] = subgroup_audit_df["Records"] >= MIN_GROUP_RECORDS
subgroup_audit_df["Serious Recall Stable Enough to Discuss"] = subgroup_audit_df["Serious Cases"] >= MIN_CLASS_CASES
subgroup_audit_df["Fatal Recall Stable Enough to Discuss"] = subgroup_audit_df["Fatal Cases"] >= MIN_CLASS_CASES

print("Subgroup performance audit (held-out test set):")
display(subgroup_audit_df.round(4))
subgroup_audit_df.to_csv(RESULTS_DIR / "subgroup_performance_audit.csv", index=False)
print("Saved: results/subgroup_performance_audit.csv")
```

### Subgroup Audit Results — Interpretation

The executed audit reports the following **held-out test** comparisons:

- **Lighting:** Daylight (**1,729 records**) had **79.12% accuracy**, **0.4173 Macro F1**, and **24.03% Serious Injury recall**. Darkness with lights lit (**692 records**) had **78.47% accuracy**, **0.4272 Macro F1**, and **32.04% Serious Injury recall**. The Serious Injury recall difference is approximately **8.0 percentage points**. Fatal Injury recall was **12.50% (2/16)** in daylight and **6.67% (1/15)** with lights lit; these fatal-case samples are too small for firm conclusions.
- **Road surface:** Dry roads (**1,833 records**) had **0.4134 Macro F1** and **26.32% Serious Injury recall**; wet or damp roads (**616 records**) had **0.4382 Macro F1** and **30.49% Serious Injury recall**. The difference in Serious Injury recall was about **4.2 percentage points**. Fatal recall was **9.52% (2/21)** versus **10.00% (1/10)**, respectively, with substantial uncertainty.
- **Weather:** Normal conditions (**1,973 records**) had **27.40% Serious Injury recall**, compared with **18.75%** during rain (**293 records**), a difference of approximately **8.7 percentage points**. Macro F1 was **0.4136** for normal weather and **0.4244** for rain. The rainy subgroup contained only **8 fatal cases**, so its **12.50% Fatal Injury recall** should not be interpreted as stable.

These are **descriptive differences, not statistically established disparities or causal effects**. Groups such as unlit darkness, snow, and fog have too few observations for reliable comparisons. The audit evaluates consistency across recorded **accident conditions**, not demographic fairness. The findings support further validation with larger Serious and Fatal Injury samples before operational use.

## 25. Examining Feature Importance of the Selected Model

Feature importance is calculated from the model selected through training cross-validation. For tree-based models, the analysis uses the fitted model's feature importance scores. If Logistic Regression is selected in a future run, the analysis uses the average absolute coefficient magnitude across classes as a descriptive alternative. These measures are model-specific and do not establish causal relationships.

```python
# Use the final tuned model selected by training cross-validation.
# This keeps feature importance consistent with the model reported in the conclusion.
best_tree_model_name = selected_config
best_tree_pipeline = tuned_models[best_tree_model_name]
fitted_preprocessor = best_tree_pipeline.named_steps["preprocessor"]
fitted_tree_model = best_tree_pipeline.named_steps["model"]

feature_names = fitted_preprocessor.get_feature_names_out()
if hasattr(fitted_tree_model, "feature_importances_"):
    importance_values = fitted_tree_model.feature_importances_
    importance_method = "Tree-based feature importance"
elif hasattr(fitted_tree_model, "coef_"):
    # Multiclass Logistic Regression: average absolute coefficients across classes.
    importance_values = np.abs(fitted_tree_model.coef_).mean(axis=0)
    importance_method = "Mean absolute standardized coefficient"
else:
    raise TypeError("The selected model has no supported feature importance method.")
encoded_importance = pd.DataFrame({
    "Encoded Feature": feature_names,
    "Importance": importance_values,
})
print("Model Used for Feature Importance:", best_tree_model_name)
print("Importance Method:", importance_method)
```

## 26. Evaluating Multiclass ROC-AUC

One-vs-rest multiclass ROC-AUC is calculated for the original untuned configurations to compare their ability to rank the severity classes. Class-level recall is also considered because ROC-AUC does not show how many Fatal Injury cases are identified at a particular decision threshold.

```python
from sklearn.metrics import roc_auc_score

roc_auc_results = []

for model_name, pipeline in fitted_models.items():

    # ROC-AUC requires predicted class probabilities.
    if hasattr(pipeline, "predict_proba"):

        # Generate predicted probabilities for each
        # accident-severity class.
        y_prob = pipeline.predict_proba(X_test)

        # Use One-vs-Rest multiclass ROC-AUC because the target
        # contains Fatal, Serious, and Slight Injury classes.
        # Macro ROC-AUC gives each severity class equal weight.
        macro_roc_auc = roc_auc_score(
            y_test,
            y_prob,
            multi_class="ovr",
            average="macro"
        )

        # Weighted ROC-AUC also considers the frequency
        # of each severity class.
        weighted_roc_auc = roc_auc_score(
            y_test,
            y_prob,
            multi_class="ovr",
            average="weighted"
        )

        roc_auc_results.append({
            "Model": model_name,
            "Macro ROC-AUC": macro_roc_auc,
            "Weighted ROC-AUC": weighted_roc_auc
        })


# Convert results to DataFrame.
roc_auc_df = pd.DataFrame(
    roc_auc_results
)

# Rank models by Macro ROC-AUC.
roc_auc_df = roc_auc_df.sort_values(
    by="Macro ROC-AUC",
    ascending=False
).reset_index(drop=True)


print("Multiclass ROC-AUC Comparison:\n")

display(
    roc_auc_df.round(4)
)
```

### ROC-AUC Results

The **untuned Balanced Random Forest** has the highest Macro ROC-AUC (**0.7012**) in this comparison, followed closely by untuned Unbalanced XGBoost (**0.6994**). These scores describe how well the models rank classes across thresholds, not how many Fatal Injury cases they correctly classify at the chosen threshold. This section compares the **original untuned models**; the final model is selected separately using tuned cross-validation Macro F1.

## 27. Grouping Feature Importance

One-hot encoded feature importance scores are combined into their original feature groups to improve interpretability. These scores describe predictive contribution within the model, not the causes of accidents.

```python
# One-hot encoding creates multiple columns from a single
# categorical predictor. Group those encoded columns back to
# their original variables so feature importance is easier
# to interpret and report in the white paper.

original_features = (
    numeric_features
    +
    categorical_features
)


def get_original_feature(
    encoded_feature
):

    if "__" in encoded_feature:

        encoded_feature = (
            encoded_feature
            .split("__", 1)[1]
        )


    if encoded_feature in original_features:

        return encoded_feature


    matches = [

        feature

        for feature
        in categorical_features

        if encoded_feature.startswith(
            feature + "_"
        )
    ]


    if matches:

        return max(
            matches,
            key=len
        )


    return encoded_feature


encoded_importance[
    "Original Feature"
] = encoded_importance[
    "Encoded Feature"
].apply(
    get_original_feature
)


feature_importance = (
    encoded_importance
    .groupby(
        "Original Feature",
        as_index=False
    )
    ["Importance"]
    .sum()
    .sort_values(
        by="Importance",
        ascending=False
    )
    .reset_index(drop=True)
)


# Display the ten most important original predictors.
display(
    feature_importance
    .head(10)
    .round(4)
)
```

### Feature Importance Results — Interpretation

For the selected **tuned Balanced Random Forest**, the leading **grouped tree-based importance** scores in the recorded analysis were **Day of Week (0.1217)**, **Hour (0.1179)**, and **Number of Vehicles Involved (0.0718)**. The importance is calculated from the **selected balanced model**, not an earlier unweighted model. `Cause_of_accident` and `Number_of_casualties` are excluded from predictors because they may encode information not available at the intended prediction point.

Tree-based importance summarizes model splitting behavior, not causation or Fatal Injury-specific predictive value. The separate permutation analysis below tests the effect of disrupting original input variables on **Macro F1**.

## 28. Top Accident Severity Predictors

Plotting the strongest feature groups makes their relative importance easier to compare. The rankings are model-specific and do not directly measure performance on rare Fatal Injury cases.

```python
top_features = (
    feature_importance
    .head(15)
    .sort_values(
        by="Importance",
        ascending=True
    )
)


plt.figure(
    figsize=(10, 7)
)


plt.barh(
    top_features[
        "Original Feature"
    ],
    top_features[
        "Importance"
    ]
)


plt.title(
    f"Top Accident Severity Predictors - {best_tree_model_name}"
)

plt.xlabel(
    "Feature Importance"
)

plt.ylabel(
    "Predictor"
)


plt.tight_layout()


plt.savefig(
    FIGURES_DIR / "feature_importance.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()

print("Saved: figures/feature_importance.png")
```

These rankings reflect how the fitted model uses the variables; they do not establish causation or guarantee reliable detection of Fatal Injury cases. The permutation analysis below provides a complementary check based on Macro F1.

### Permutation Feature Importance Using Macro F1

Tree-based feature importance measures how the fitted model uses predictors in its splits. This can favor variables with many possible values and does not directly show whether a predictor improves classification performance. As a complementary check, **permutation importance** measures how much the selected model’s **Macro F1** decreases when one original input column is randomly shuffled while all other columns remain unchanged.

The analysis uses the **selected tuned pipeline** on the held-out test set, without refitting or selecting a different model. It is a **post-selection interpretability diagnostic**, not a basis for choosing or tuning the model. Because the test set is used for interpretation, these results are exploratory; an independent validation dataset would be preferable for confirmatory interpretation. Each variable is shuffled five times to show variability. The original columns are shuffled before one-hot encoding, so categorical variables remain grouped automatically.

A large positive decrease suggests that the model relies on that variable for overall class-balanced classification. Values near zero or below zero indicate little or unstable benefit in this test sample. **Macro F1 is not specific to Fatal Injury**, so even an important predictor may not help identify rare fatal cases. Importance is not evidence of causation.

```python
# Model-agnostic permutation importance for the selected, tuned pipeline.
# This is an exploratory test-set interpretation, NOT model selection or tuning.
from sklearn.inspection import permutation_importance
from sklearn.metrics import make_scorer, f1_score

permutation_results = permutation_importance(
    best_pipeline,
    X_test,
    y_test,
    scoring=make_scorer(f1_score, average="macro", zero_division=0),
    n_repeats=5,
    random_state=RANDOM_STATE,
    n_jobs=1,
)

permutation_importance_df = (
    pd.DataFrame({
        "Original Feature": X_test.columns,
        "Mean Macro F1 Decrease": permutation_results.importances_mean,
        "Std. Deviation": permutation_results.importances_std,
    })
    .sort_values("Mean Macro F1 Decrease", ascending=False)
    .reset_index(drop=True)
)

print("Selected model:", best_model_name)
print("Metric: decrease in held-out test Macro F1 after feature permutation")
display(permutation_importance_df.head(15).round(4))
permutation_importance_df.to_csv(
    RESULTS_DIR / "permutation_feature_importance.csv", index=False
)

# Visualize the most influential original predictors.
top_permutation = permutation_importance_df.head(15).iloc[::-1]
fig, ax = plt.subplots(figsize=(10, 7))
ax.barh(
    top_permutation["Original Feature"],
    top_permutation["Mean Macro F1 Decrease"],
    xerr=top_permutation["Std. Deviation"],
    capsize=3,
)
ax.axvline(0, color="gray", linewidth=1)
ax.set_xlabel("Decrease in Macro F1 when feature is shuffled")
ax.set_ylabel("Original predictor")
ax.set_title("Permutation Importance — Selected Tuned Model")
plt.tight_layout()
plt.savefig(FIGURES_DIR / "permutation_feature_importance.png", dpi=300, bbox_inches="tight")
plt.show()
print("Saved: results/permutation_feature_importance.csv")
print("Saved: figures/permutation_feature_importance.png")
```

### Interpreting Permutation Importance — Recorded Results

The recorded permutation analysis of the **selected tuned Balanced Random Forest** found the largest mean decreases in **Macro F1** when shuffling **Number of Vehicles Involved (0.0568)**, **Day of Week (0.0261)**, **Types of Junction (0.0237)**, and **Hour (0.0206)**. Weather Conditions produced a much smaller mean decrease (**0.0018**). The rankings differ from tree-based importance because the two methods measure different properties.

The importance estimates are exploratory, are affected by correlations among predictors, and are based on the held-out test data. They do not establish causality or demonstrate that a feature reliably identifies Fatal Injury cases. The selected model's low Fatal Injury recall remains a central limitation.

## 31. Assumptions

The analysis assumes that recorded accident severity and categorical descriptions are sufficiently consistent for comparison. Median and most-frequent-value imputation assume that the available information provides a reasonable basis for filling missing predictor values. Stratified sampling preserves the observed class proportions but does not guarantee that future accidents will follow the same patterns. The analysis also assumes that excluded post-accident information should not be treated as available predictors in a prospective setting.

## 32. Limitations

Fatal Injury cases account for only about **1.28%** of the dataset, limiting the reliability of minority-class predictions and subgroup recall estimates. The records are from Addis Ababa during **2017–2020**, so the results cannot be generalized to other places or periods without additional validation. Missing data, unmeasured exposure such as traffic volume, and the retrospective nature of some accident variables further limit interpretation. Feature importance and statistical associations do not establish causation.

## 33. Challenges

The central challenge was class imbalance: models could achieve relatively high accuracy while missing most Fatal Injury cases. The workflow therefore compared balanced and unbalanced training, used stratified cross-validation, tuned configurations with Macro F1, and examined class-specific recall. Missing categorical values and one-hot encoding required consistent preprocessing. Potential target leakage was addressed by excluding casualty-related and investigation-dependent fields.

## 34. Future Uses

With a larger and more representative dataset, the workflow could support exploratory road-safety dashboards, analysis of accident patterns, and research into severity-related factors. Future studies could add traffic volume, speed, location, roadway design, and additional years of records. Further work should examine sampling strategies, probability calibration, class-specific decision thresholds, and external validation before considering any operational application.

## 35. Recommendations

Road-safety stakeholders should treat these results as exploratory evidence rather than an automated decision system. Future model evaluation should prioritize **Fatal and Serious Injury recall**, Macro F1, balanced accuracy, and error patterns—not accuracy alone. The low Fatal Injury recall indicates that the selected model is not suitable for safety-critical decisions in its current form. Additional severe-accident records and independent validation are priorities.

## 36. Implementation Plan

1. **Data governance:** Confirm the legal basis for using new accident records, document definitions, and establish quality checks.
2. **Reproducible preparation:** Apply the same leakage exclusions and training-fitted preprocessing to updated data.
3. **Independent validation:** Evaluate the model on later years and, if possible, a different geographic area; assess subgroup consistency and minority-class recall.
4. **Model improvement:** Investigate imbalance methods, probability calibration, and threshold choices using validation data only.
5. **Expert review:** Have transportation and public-safety specialists assess error costs and appropriate use cases.
6. **Controlled pilot only if validated:** Start with human-reviewed analytical reporting, monitor performance drift, and define clear stop criteria. No automated safety-critical deployment is recommended from the current results.

## 37. Ethical Assessment

**Identify.** Key concerns include missed Fatal and Serious Injury cases, incomplete accident records, privacy risks if records are linked to individuals, geographic bias, and the misuse of correlation or feature importance as causal evidence.

**Assess.** The selected model’s **9.68% Fatal Injury recall** is a concrete performance concern. The executed subgroup audit found Serious Injury recall of **24.03% in daylight (n=1,729)** versus **32.04% with lights lit at night (n=692)**, and **27.40% in normal weather (n=1,973)** versus **18.75% in rain (n=293)**. These descriptive differences warrant further testing; rare Fatal Injury counts make subgroup fatal recall highly uncertain. These operational subgroups are not substitutes for demographic fairness analysis.

**Mitigate.** The project uses leakage-aware predictors, stratified validation, class-sensitive metrics, the final-model confusion matrix, a subgroup audit, and a Macro F1 permutation-importance diagnostic. Before any real-world use, the model would require more representative data, external validation, subgroup review with sufficient cases, privacy controls, and human oversight. Its current performance does not support automated safety-critical decisions.

**Data sourcing and responsible interpretation.** The analysis uses a publicly available historical accident dataset. Results are presented as retrospective associations and predictive patterns, not proof of causality or readiness for deployment.
